```python
import torch
import matplotlib.pyplot as plt

from scripts.config import AGGREGATE, USE_ATTENTION
from scripts.train_classic import train
from scripts.train_kfold import train_kfold
from scripts.generate_submission_classic import generate_submission
from scripts.generate_submission_multi_models import generate_submission_from_models
```

```python
training_type = "classic"  # options: "classic", "kfold"
```

```python
if training_type == "classic":
    train()
elif training_type == "kfold":
    train_kfold()
```

```python
if training_type == "classic":
    paths = {
        1 : {'model' : f'output/training_logs_aggregate_{AGGREGATE}_attn_{USE_ATTENTION}.pth',
            'aggregate' : AGGREGATE,
            'use_attention' : USE_ATTENTION},
    }
elif training_type == "kfold":
    paths = {
        1 : {'model' : f'output/training_logs_fold_0_aggregate_{AGGREGATE}_attn_{USE_ATTENTION}.pth',
            'aggregate' : AGGREGATE,
            'use_attention' : USE_ATTENTION},
        2 : {'model' : f'output/training_logs_fold_1_aggregate_{AGGREGATE}_attn_{USE_ATTENTION}.pth',
            'aggregate' : AGGREGATE,
            'use_attention' : USE_ATTENTION},
        3 : {'model' : f'output/training_logs_fold_2_aggregate_{AGGREGATE}_attn_{USE_ATTENTION}.pth',
            'aggregate' : AGGREGATE,
            'use_attention' : USE_ATTENTION},
        4 : {'model' : f'output/training_logs_fold_4_aggregate_{AGGREGATE}_attn_{USE_ATTENTION}.pth',
            'aggregate' : AGGREGATE,
            'use_attention' : USE_ATTENTION},
        5 : {'model' : f'output/training_logs_fold_4_aggregate_{AGGREGATE}_attn_{USE_ATTENTION}.pth',
            'aggregate' : AGGREGATE,
            'use_attention' : USE_ATTENTION},
    }
    
    colors = {
        1: "#1f77b4",
        2: "#ff7f0e",
        3: "#2ca02c",
        5: "#d62728",
        4: "#242323",
    }

    logs_dict = {}

    for key in paths:
        logs = torch.load(paths[key]['model'])
        logs_dict[key] = logs
```

```python
plt.figure(figsize=(10, 6))
for k, logs in logs_dict.items():
    lw = 3 if k == 1 else 1.8
    plt.plot(logs['epoch'], logs['train_loss'], 
             label=f"Model {k}", 
             linewidth=lw, 
             color=colors[k])
plt.title("Training Loss Curves")
plt.xlabel("Epoch")
plt.ylabel("Train Loss")
plt.grid(alpha=0.25)
plt.legend()
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
for k, logs in logs_dict.items():
    lw = 3 if k == 1 else 1.8
    plt.plot(logs['epoch'], logs['val_score'], 
             label=f"Model {k}",
             linewidth=lw, 
             color=colors[k])
plt.title("Validation Score Curves")
plt.xlabel("Epoch")
plt.ylabel("Validation Score")
plt.grid(alpha=0.25)
plt.legend()
plt.tight_layout()
plt.show()
```
