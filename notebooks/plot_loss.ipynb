```python
import torch
import matplotlib.pyplot as plt

paths = {
    1 : {'model' : 'training_logs_aggregate_True_attn_True.pth',
         'aggregate' : True,
         'use_attention' : True},
    2 : {'model' : 'training_logs_aggregate_True_attn_False.pth',
         'aggregate' : True,
         'use_attention' : False},
    3 : {'model' : 'training_logs_aggregate_False_attn_True.pth',
         'aggregate' : False,
         'use_attention' : True},
    4 : {'model' : 'training_logs_aggregate_False_attn_False.pth',
         'aggregate' : False,
         'use_attention' : False}
}
```

```python
colors = {
    1: "#1f77b4",
    2: "#ff7f0e",
    3: "#2ca02c",
    4: "#d62728"
}

logs_dict = {}

for key in paths:
    logs = torch.load(paths[key]['model'])
    logs_dict[key] = logs
```

```python

plt.figure(figsize=(10, 6))
for k, logs in logs_dict.items():
    lw = 3 if k == 1 else 1.8
    plt.plot(logs['epoch'], logs['train_loss'], 
             label=f"Model {k}" + (" (overfitted)" if k == 3 else ""), 
             linewidth=lw, 
             color=colors[k])
plt.title("Training Loss Curves")
plt.xlabel("Epoch")
plt.ylabel("Train Loss")
plt.grid(alpha=0.25)
plt.legend()
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(10, 6))
for k, logs in logs_dict.items():
    lw = 3 if k == 1 else 1.8
    plt.plot(logs['epoch'], logs['val_score'], 
             label=f"Model {k}" + (" (overfitted)" if k == 3 else ""), 
             linewidth=lw, 
             color=colors[k])
plt.title("Validation Score Curves")
plt.xlabel("Epoch")
plt.ylabel("Validation Score")
plt.grid(alpha=0.25)
plt.legend()
plt.tight_layout()
plt.show()
```
