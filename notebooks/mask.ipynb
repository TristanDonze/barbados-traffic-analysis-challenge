```python
import sys
import os
import cv2
import matplotlib.pyplot as plt

notebook_dir = os.path.dirname(os.path.abspath("__file__"))
src_dir = os.path.abspath(os.path.join(notebook_dir, "../src"))
sys.path.append(src_dir)

from features.mask import all_masks
```

```python
path_1 = "../src/mask/extracted_images/normanniles1_first_frame.jpg"
path_2 = "../src/mask/extracted_images/normanniles2_first_frame.jpg"
path_3 = "../src/mask/extracted_images/normanniles3_first_frame.jpg"
path_4 = "../src/mask/extracted_images/normanniles4_first_frame.jpg"

first_frames = {
    "normanniles1": cv2.imread(path_1),
    "normanniles2": cv2.imread(path_2),
    "normanniles3": cv2.imread(path_3),
    "normanniles4": cv2.imread(path_4),
}
```

```python
def apply_mask(image, mask):
    """Apply a binary mask to an image with red color and 50% opacity."""
    red_overlay = image.copy()
    red_overlay[mask > 0] = [0, 0, 255]

    blended = cv2.addWeighted(image, 0.5, red_overlay, 0.5, 0)
    return blended
```

```python
i = 4
img_np = apply_mask(first_frames[f"normanniles{i}"], all_masks[f"normanniles{i}"][3])
plt.imshow(cv2.cvtColor(img_np, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```
