```python
import os
import pandas as pd

train_data_file_path = "../data/Train.csv"
test_segments_file_path = "../data/TestInputSegments.csv"
videos_root_directory = "../data/videos/"

training_dataframe = pd.read_csv(train_data_file_path)
test_segments_dataframe = pd.read_csv(test_segments_file_path)
```

```python
train_video_paths_set = set()
for relative_video_path in training_dataframe['videos']:
    full_path = os.path.join(videos_root_directory,
                            relative_video_path)
    real_video_system_path = os.path.realpath(full_path)
    train_video_paths_set.add(real_video_system_path)

print(f"Total number of videos in Train.csv : {len(train_video_paths_set)}")
```

```python
test_video_paths_set = set()
for relative_video_path in test_segments_dataframe['videos']:
    full_path = os.path.join(videos_root_directory, relative_video_path)
    real_video_system_path = os.path.realpath(full_path)
    test_video_paths_set.add(real_video_system_path)

print(
    f"Total number of videos in TestInputSegments.csv : {len(test_video_paths_set)}")
```

```python
physical_video_files_set = set()

for root, dirs, files in os.walk(videos_root_directory):
    for dir_ in dirs:
        real_dir_path = os.path.realpath(os.path.join(root, dir_))
        for video_file in os.listdir(real_dir_path):
            full_path = os.path.realpath(os.path.join(real_dir_path, video_file))
            real_video_system_path = os.path.realpath(full_path)
            if real_video_system_path.endswith('.mp4'):
                physical_video_files_set.add(real_video_system_path)

print(f"Total number of videos on disk : {len(physical_video_files_set)}")
```

```python
# print first 10 train set video paths
print("First 10 video paths in Train.csv:")
for i, video_path in enumerate(train_video_paths_set):
    if i >= 10:
        break
    print(video_path)
```

```python
videos_in_both_train_and_test = train_video_paths_set.intersection(
    test_video_paths_set)
train_videos_missing_from_disk = train_video_paths_set.difference(
    physical_video_files_set)
test_videos_missing_from_disk = test_video_paths_set.difference(
    physical_video_files_set)
physical_videos_not_in_any_csv = physical_video_files_set.difference(
    train_video_paths_set.union(test_video_paths_set))
test_videos_present_on_disk_but_not_in_train = (test_video_paths_set.intersection(
    physical_video_files_set)).difference(train_video_paths_set)
train_test_intersection = train_video_paths_set.intersection(
    test_video_paths_set)

print(f"Total unique videos in Train CSV: {len(train_video_paths_set)}")
print(f"Total unique videos in Test CSV: {len(test_video_paths_set)}")
print(f"Total unique videos found on disk: {len(physical_video_files_set)}")
print("-" * 60)
print(
    f"Overlap: Videos present in BOTH Train and Test CSVs: {len(videos_in_both_train_and_test)}")
print(
    f"Integrity: Train videos missing from disk: {len(train_videos_missing_from_disk)}")
print(
    f"Integrity: Test videos missing from disk: {len(test_videos_missing_from_disk)}")
print(
    f"Cleanliness: Videos on disk but NOT used in Train or Test: {len(physical_videos_not_in_any_csv)}")
print(
    f"Dataset Split: Test videos present on disk but NOT in Train: {len(test_videos_present_on_disk_but_not_in_train)}")
print(
    f"Train/Test Intersection: Videos present in BOTH Train and Test CSVs: {len(train_test_intersection)}"
)
```
